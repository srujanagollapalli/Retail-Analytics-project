```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("Retail_Analytics_Sample_Data.xlsx")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.describe()
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
highest_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("Highest Sales Region:", highest_region)
print("Sales:", round(highest_region_sales, 2))
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
profit_by_category = df.groupby("Category")["Profit"].sum()

print(profit_by_category)
```

```python
highest_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

print("Highest Sales Category:", highest_category)
print("Sales:", round(highest_category_sales, 2))
```

```python
highest_profit_category = profit_by_category.idxmax()
highest_category_profit = profit_by_category.max()

print("Highest Profit Category:", highest_profit_category)
print("Profit:", round(highest_category_profit, 2))
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print(product_sales)
```

```python
top_products = product_sales.head(5)

print(top_products)
```

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
monthly_profit = df.groupby("Month")["Profit"].sum()

print(monthly_profit)
```

```python
region_sales.plot(kind="bar", figsize=(8,5))

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
category_sales.plot(kind="bar", figsize=(8,5))

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
category_data = df.groupby("Category")[["Sales", "Profit"]].sum()

category_data.plot(kind="bar", figsize=(8,5))

plt.title("Sales vs Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
top_products.plot(kind="bar", figsize=(8,5))

plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_sales.plot(kind="line", marker="o", figsize=(9,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
monthly_profit.plot(kind="line", marker="o", figsize=(9,5))

plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
category_quantity = df.groupby("Category")["Quantity"].sum()

category_quantity.plot(kind="bar", figsize=(8,5))

plt.title("Quantity Sold by Category")
plt.xlabel("Category")
plt.ylabel("Quantity")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid()
plt.show()
```
